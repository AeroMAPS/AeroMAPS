# Example for using climate models

This notebook aims at using the climate model class of AeroMAPS, which relies on the use of the [AeroCM](https://github.com/AeroMAPS/AeroCM) framework.

## Import librairies and configuration

```python
# --- Import libraries ---
%matplotlib widget
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np

# --- AeroMPAS processes ---
from aeromaps import create_process
```

## Set models

The choice of the models is directly managed through the use of configuration files provided in `./data`.

On the one hand, some basic models from AeroMAPS are used, in addition to the default energy carriers models. Note that two specific AeroMAPS models are also studied alternatively: the `FuelEffectCorrectionContrails` model which allows taking into account the effects of alternative fuels on contrails, and the `WithoutFuelEffectCorrectionContrails` one which does not allow them.

On the other hand, the climate models are considered using the `ClimateModel` class. Currently, 4 aviation climate models are available: IPCC, GWP*, LWE, FaIR. Default settings for the different climate models are available, but the user can change them easily by modifying the yaml climate model data file. 

## Create process and computation

This example notebook uses the default AeroMAPS inputs data.

We are creating 6 processes, corresponding to 6 configuration files:
- 4 processes correspond to the different climate models with default settings, coupled to the use of the `FuelEffectCorrectionContrails` model
- 1 process correspond to the FaIR climate model with default settings, coupled to the use of the `WithoutFuelEffectCorrectionContrails` model
- 1 process is a modified version of the FaIR climate model by assuming a lower climate sensitivity to contrails, coupled to the use of the `FuelEffectCorrectionContrails` model

```python
process1 = create_process(configuration_file="data/config_fair_kerosene.yaml")
process1.parameters.karcher_curve_case = "Low vPM"
process1.parameters.operations_contrails_final_gain = 0.0  # [%]
process1.parameters.operations_contrails_final_overconsumption = 0.0  # [%]
process1.parameters.operations_contrails_start_year = 2030
process1.parameters.operations_contrails_duration = 15.0
process1.compute()
```

## Results

The user can then display the results for each configuration (e.g. default FaIR with the effects of alternative fuels on contrails).

We can for instance compare the results obtained for the different configurations.

```python
fig = plt.figure(figsize=(10, 6))
gs = gridspec.GridSpec(2, 2, width_ratios=[1.6, 1])
start_year = 1940
end_year = 2050
years_data = range(start_year, end_year+1)

ax1 = fig.add_subplot(gs[:, 0])
ax1.plot(years_data, 1000 * process1.data["climate_outputs"][["temperature_increase_from_aviation"]], label='Default - Baseline', color='C0')
ax1.plot(years_data, 1000 * process2.data["climate_outputs"][["temperature_increase_from_aviation"]], label='Default - With SAF', color='C0', linestyle="--")
ax1.plot(years_data, 1000 * process3.data["climate_outputs"][["temperature_increase_from_aviation"]], label='Default - With contrail avoidance', color='C0', linestyle=":")
ax1.plot(years_data, 1000 * process4.data["climate_outputs"][["temperature_increase_from_aviation"]], label='High vPM - Baseline', color='C1')
ax1.plot(years_data, 1000 * process5.data["climate_outputs"][["temperature_increase_from_aviation"]], label='High vPM - With SAF', color='C1', linestyle="--")
ax1.plot(years_data, 1000 * process6.data["climate_outputs"][["temperature_increase_from_aviation"]], label='High vPM - With contrail avoidance', color='C1', linestyle=":")
ax1.plot(years_data, 1000 * process7.data["climate_outputs"][["temperature_increase_from_aviation"]], label='New species settings - Baseline', color='C2')
ax1.plot(years_data, 1000 * process8.data["climate_outputs"][["temperature_increase_from_aviation"]], label='New species settings - With SAF', color='C2', linestyle="--")
ax1.plot(years_data, 1000 * process9.data["climate_outputs"][["temperature_increase_from_aviation"]], label='New species settings - With contrail avoidance', color='C2', linestyle=":")
ax1.legend()
ax1.grid()
ax1.set_xlim(start_year, end_year)
ax1.set_xlabel('Year')
ax1.set_ylabel('Surface temperature change [mK]')
ax1.set_title("Aviation total climate impacts")

ax2 = fig.add_subplot(gs[0, 1])
ax2.plot(years_data, 1000 * process1.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C0')
ax2.plot(years_data, 1000 * process2.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C0', linestyle="--")
ax2.plot(years_data, 1000 * process3.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C0', linestyle=":")
ax2.plot(years_data, 1000 * process4.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C1')
ax2.plot(years_data, 1000 * process5.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C1', linestyle="--")
ax2.plot(years_data, 1000 * process6.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C1', linestyle=":")
ax2.plot(years_data, 1000 * process7.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C2')
ax2.plot(years_data, 1000 * process8.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C2', linestyle="--")
ax2.plot(years_data, 1000 * process9.data["climate_outputs"][["temperature_increase_from_co2_from_aviation"]], color='C2', linestyle=":")
ax2.grid()
ax2.set_xlim(start_year, end_year)
ax2.set_xlabel('Year')
ax2.set_ylabel('Surface temperature change [mK]')
ax2.set_title('Aviation CO₂ climate impacts')

ax3 = fig.add_subplot(gs[1, 1])
ax3.plot(years_data, 1000 * process1.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C0')
ax3.plot(years_data, 1000 * process2.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C0', linestyle="--")
ax3.plot(years_data, 1000 * process3.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C0', linestyle=":")
ax3.plot(years_data, 1000 * process4.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C1')
ax3.plot(years_data, 1000 * process5.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C1', linestyle="--")
ax3.plot(years_data, 1000 * process6.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C1', linestyle=":")
ax3.plot(years_data, 1000 * process7.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C2')
ax3.plot(years_data, 1000 * process8.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C2', linestyle="--")
ax3.plot(years_data, 1000 * process9.data["climate_outputs"][["temperature_increase_from_non_co2_from_aviation"]], color='C2', linestyle=":")
ax3.grid()
ax3.set_xlim(start_year, end_year)
ax3.set_xlabel('Year')
ax3.set_ylabel('Surface temperature change [mK]')
ax3.set_title('Aviation non-CO₂ climate impacts')

plt.tight_layout()
plt.show()
```
