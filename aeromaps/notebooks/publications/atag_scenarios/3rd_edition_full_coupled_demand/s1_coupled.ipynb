# Demand-price coupling of ATAG Waypoint S1 scenario




```python
%matplotlib widget


# The scenario ships with the package; copy it somewhere writable before running,
# so this notebook's outputs and regenerated inputs land in ./workdir rather than
# in the installed AeroMAPS.
from aeromaps.utils.scenarios import prepare_scenario
from aeromaps import create_process
from get_data import get_ar6_input_data

SCENARIO = prepare_scenario("atag_3rd_edition_coupled_demand")

ar6_data, ar6_years = get_ar6_input_data(start_year=2010, end_year=2100, plot_data=True)
```

## Background scenario

To start, we'll use the SSP2-3.4 scenario, representing a scenario with moderate climate ambition (~2.5°C temp. increase by 2100).


```python
background_scenario = "SSP2-26"
process = create_process(str(SCENARIO / "config_files" / "config_s1.yaml"))
process.parameters.population_reference_years = ar6_years
process.parameters.population_reference_years_values = ar6_data["population"][background_scenario]
process.parameters.gdp_per_capita_reference_years = ar6_years
process.parameters.gdp_per_capita_reference_years_values = ar6_data["gdp_per_capita"][
    background_scenario
]
process.parameters.carbon_tax_reference_years = ar6_years
process.parameters.carbon_tax_reference_years_values = ar6_data["carbon_tax"][background_scenario]
process.parameters.exogenous_carbon_price_reference_years = ar6_years
process.parameters.exogenous_carbon_price_reference_years_values = ar6_data["carbon_tax"][
    background_scenario
]
process.compute()
```
