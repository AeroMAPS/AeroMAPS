## Load modules

First, the user has to load the framework and generate a process.

```python
# --- Import libraries ---
%matplotlib widget
import matplotlib.pyplot as plt
import time
import sys

sys.path.insert(0, "./utils/")
from plots import (
    plot_stacked_evolution_subplots,
    stacked_cascade_scenarios
)
from aeromaps import create_process

plt.style.use("bmh")
```

## Scenario IS0

```python
# --- Set AeroMAPS models and create process for scenario IS0 ---
# Note: first call to LCA module takes several dozen minutes depending on CPU, as it will install ecoinvent/premise databases (unless previsouly installed)
# and parametrize the LCA model declared in LCA configuration file (each time the kernel is restarted).
process = create_process(
    configuration_file="./data/config_files/lca/config_is0medium_lca_advanced.yaml",
)
```

```python
# --- Run assessment ---
start_time = time.time()
process.compute()
process.write_json()
print("--- %s seconds ---" % (time.time() - start_time))
```

```python
# --- Outputs ---
process_data_vector_outputs_scenario_0 = process.data["vector_outputs"]
process_data_float_inputs_scenario_0 = process.data["float_inputs"]
process_data_climate_scenario_0 = process.data["climate_outputs"]
lca_outputs_scenario_0 = process.data["lca_outputs"]
lca_outputs_scenario_0
```

```python
# --- Plot results ---
plot_stacked_evolution_subplots(lca_outputs_scenario_0, end_year=2050)
```

## Summary of LCA results

```python
# --- Bar plot evolution through scenarios ---
stacked_cascade_scenarios(
    {
        "International Aviation\n2019 (ref.)": [lca_outputs_scenario_0, 2019],
        "Integrated Scenario 1\n2050": [lca_outputs_scenario_1, 2050],
        "Integrated Scenario 2\n2050": [lca_outputs_scenario_2, 2050],
        "Integrated Scenario 3\n2050": [lca_outputs_scenario_3, 2050],
    },
    ylim = [-4e5, 4.5e6], # y-axis limits for short-term impacts
    #ylim = [-6e5, 1.3e7],  # y-axis limits for long-term impacts
);
#plt.savefig('fig-ltag_scenarios_HH_short_term_cc.pdf')
#plt.savefig('fig-ltag_scenarios_HH_short_and_long_term_cc.pdf')
```

```python
# --- Bar plot for 2019 (comparison of lcia methods) ---

lcia_methods_dict = {
    'LCIA Aviation': (
        "This work", 
        [
            "climate change: human health",
            "pm25: human health",
            "o3: human health",
        ]
    ),
    'ReCiPe': (
        "ReCiPe2016\n& Lee et al. (2021)",
        [
            "particulate matter formation",
            "photochemical oxidant formation: human health",
            "climate change: human health",
        ]
    ),
    'IWP': (
        "Impact World+\n& Lee et al. (2021)",
        [
            "particulate matter formation",
            "photochemical oxidant formation: human health",
            "climate change: human health",
        ]
    ),
}
       
    
fig = stacked_cascade_scenarios(
    {
        method: (lca_outputs_scenario_0, 2019)
        for method in lcia_methods_dict
    },
    ylim=[-3e5, 3e6],
    comparison='methods',
    methods_dict=lcia_methods_dict,
)

# Add arrow for ReCiPe O3
ax = fig.axes[1]
o3_bar_top = 1.28e6
label_x = 2.0
label_y = 1.41e6
ax.annotate(
    "",
    xy=(2.0, o3_bar_top),      
    xytext=(label_x, label_y),   
    arrowprops=dict(
        arrowstyle="->",
        lw=1.2,
        color="black",
    )
)

fig.savefig('fig-ltag_scenarios_HH_LCIA_comparison.pdf')
plt.show()
```
