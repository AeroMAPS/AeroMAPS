# Basic verification of all plots for a single AeroMAPS scenario

## Load, create process and compute

```python
%matplotlib widget
from aeromaps import create_process
```

```python
process = create_process(configuration_file="../tested_configs/config_advanced.yaml")
process.compute()
```

## Plot all available plots

```python
for plot_name in process.list_available_plots():
    if "simple" not in plot_name:
        try:
            process.plot(plot_name, save=False)
        except Exception as e:
            print(f"{plot_name}: {e}")
```

```python
process_simple = create_process(
    configuration_file="../tested_configs/config_advanced_simplified.yaml"
)
process_simple.compute()

for plot_name in process_simple.list_available_plots():
    if "simple" in plot_name:
        try:
            process_simple.plot(plot_name, save=False)
        except Exception as e:
            print(f"{plot_name}: {e}")
```
